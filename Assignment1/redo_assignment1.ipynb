```python
from Risk_Library import first_four_moments
```

```python
f4m = first_four_moments(problem1, method="unbiased", kurtosis_type="excess")
print(f4m.loc["Skewness", "x"])
print(f4m)
```

```python
#Question 1.b

from scipy.stats import norm

mu = f4m.loc["Mean", "x"]
sigma = np.sqrt(f4m.loc["Variance", "x"])

norm_fit = norm(loc=mu, scale=sigma)
```

```python
q01 = norm_fit.ppf(0.01)

sample = problem1["x"]
n_below = ( sample< q01).sum()
n_expected = 0.01 * len(sample)

print(f"Normal 1% quantile = {q01}")
print(f"Observations below quantile = {n_below}")
print(f"Expected observations = {n_expected}")
```
