```python
import numpy as np
import pandas as pd
```

```python
#Q1

problem1 = pd.read_csv("problem1.csv")
print(problem1.head())

from Risk_Library import first_four_moments
f4m = first_four_moments(problem1)
print(f4m)

mu = f4m.loc["Mean","X"]
sigma = np.sqrt(f4m.loc["Variance","X"])
```

```python
#Q1.3

from Risk_Library import fit_normal_distribution, fit_t_distribution

data = problem1["X"]
norm_fit = fit_normal_distribution(data, output="params")
t_fit = fit_t_distribution(data, output="params")

from Risk_Library import t_distribution_aicc
t_aicc = t_distribution_aicc(data, t_fit)
print(f"t_aic: {t_aicc}")

from scipy import stats
from utils import aicc
normal_loglik = stats.norm.logpdf(data, loc=norm_fit["mu"], scale=norm_fit["sigma"]).sum()
norm_aicc = aicc(normal_loglik, num_params=2, num_obs=len(data))
print(f"normal_aic: {norm_aicc}")


```

```python
#Q1.4
from Risk_Library import calculate_var, calculate_es

#5% alpha
norm_var_5 = calculate_var(data, alpha= 0.05)
t_var_5 = calculate_var(data, distribution= "t", alpha= 0.05)
print(f"Normal VaR with alpha= 5%: {norm_var_5["VaR Absolute"]}")
print(f"T VaR with alpha= 5%: {t_var_5["VaR Absolute"]}")

#......................then cal 5% alpha es

#......................do same for 1% alpha
```
